### 3.2. Métodos de interés en NumPy
- Generador: `arange()`
- Shape: Forma (filas x columnas)
- Reshape: Cambiar forma
- Slicing: Rebanada. Quedarnos con una o varias filas o una o varias columnas de la matriz total.
- Aplanar: Cambiar a una dimensión
- Ravel: equivalente a `flatten()` pero trabaja con vistas.
- Reducción: sum, prod y diff

```python
import numpy as np
# Generamos array
# Con lista...
a = [ x for x in range(30)]
# Con arange
a = np.arange(30)
forma_a = np.shape(a)

print(f"Mostramos 'a' {forma_a}:\n", a)


# Cambiando la forma
# 2 filas 10 columnas
b = a.reshape(3, 10)
forma_b = b.shape
print(f"\nMostramos 'b' {forma_b}:\n", b)

#---
print("-"*30)
c = b.reshape(10, 3) + np.array([1, 2, 3])
print("Suma de b:", c)
print("-"*30)


# 10 filas 2 columnas
c = b.reshape(10, 3)
forma_c = c.shape
print(f"\nMostramos 'c' {forma_c}:\n", c)

# Incluso a una dimensión 20 x 1
d = c.reshape(30)
print(f"\nMostramos 'd'", d)

# Rabanadas
# Primera fila (todas las columnas)
fila = c[0,:]
print("\nPrimera fila:", fila)

# Del 3º al 5 elemento (incluído)
algunos = d[2:5]
print("Elementos 3 a 5 (inc):\n", c) 

# De tercera a 5 (incluída) filas (y todas las columnas)
cacho = c[2:5, :]
print("Filas 3 a 5 (inc):\n", cacho)

# Aplanado
aplanado = c.flatten()
print("\nAplanado: ", aplanado)

# Reducción
b = np.array([1,2,3,4]).reshape(2,2)
print(f"Nuevo b:\n", b)
suma = b.sum()
print("Suma 'b' todos los elementos: ", suma)
fila_de_suma_por_columna = b.sum(axis=0)
print("Suma 'b' por columnas (me quedo una fila): ", fila_de_sumas_por_columna)
columnas_de_sumas_por_fila = b.sum(axis=1)
print("Suma 'b' por filas (me quedo una columna): ", columnas_de_sumas_por_fila)
producto = b.prod(axis=1)
print("Producto de 'b': ", producto)
distancia = np.diff(a)

print("Distancia en tramos 'b': ", distancia)
```

Los **datos** de entrada:

```python
# Datos de entrada: Lista de diccionarios
ventas = [
    {"id": 1, "producto": "Laptop", "cantidad": 2, "precio_unitario": 850.0, "descuento": 0.10},
    {"id": 2, "producto": "Ratón", "cantidad": 15, "precio_unitario": 25.5, "descuento": 0.05},
    {"id": 3, "producto": "Teclado", "cantidad": 8, "precio_unitario": 45.0, "descuento": 0.00},
    {"id": 4, "producto": "Monitor", "cantidad": 4, "precio_unitario": 300.0, "descuento": 0.15}
]
```

```python
import numpy as np

def calcular_media_ventas(lista):
    # Arrays unidimensionales paralelos (simulando columnas de una tabla)
    cantidades = np.array([v["cantidad"] for v in lista])
    precios = np.array([v["precio_unitario"] for v in lista])
    descuentos = np.array([v["descuento"] for v in lista])
    

    # 1. Operación vectorizada elemento a elemento (SIN BUCLES EXPLÍCITOS)
    totales_numpy = cantidades * precios * (1 - descuentos)

    print("Totales por venta (NumPy vectorizado):", totales_numpy)

    # 2. Uso de Broadcasting y funciones de agregación integradas
    # O como multiplicar por un escalar
    iva = totales_numpy * 0.21
    totales_con_iva = totales_numpy + iva
    print("Iva y Totales con IVA (Broadcasting):", iva, totales_con_iva)

    # Cálculos estadísticos directos sin bucles
    media_numpy = totales_numpy.mean()
    desviacion_numpy = totales_numpy.std()
    maximo_venta = totales_numpy.max()

    print(f"Media: {media_numpy:.2f} € | Desviación estándar: {desviacion_numpy:.2f} € | Máximo: {maximo_venta:.2f} €")
    
    return media_numpy
```

#### Ejercicio 3.3.3: Normalización Min-Max (CE1.e)
Implementa la función `normalizar_min_max(arr)` para escalar un array 1D al rango $[0, 1]$:
$$X_{norm} = \frac{X - X_{min}}{X_{max} - X_{min}}$$
*(Si $X_{max} == X_{min}$, retorne un array de ceros).*

```python
import numpy as np

def normalizar_min_max(arr: np.ndarray) -> np.ndarray:
    if arr.min() == arr.max():
            return np.zeros_like(arr)
            # tambien vale return arr * 0
    else:
        return (arr - arr.min())/(arr.max() - arr.min())
    raise NotImplementedError("Implementa la función normalizar_min_max")
```

```python
%%ipytest -rA

def test_normalizar_min_max():
    
    datos = np.array([10.0, 20.0, 30.0, 40.0, 50.0])
    resultado = normalizar_min_max(datos)
    esperado = np.array([0.0, 0.25, 0.5, 0.75, 1.0])

    np.testing.assert_allclose(resultado, esperado, rtol=1e-5)
    assert normalizar_min_max(np.array([5, 5, 5])).sum() == 0, "Error en valores constantes"
    print("✅ Test Ejercicio 3.3.3 superado.")
```
